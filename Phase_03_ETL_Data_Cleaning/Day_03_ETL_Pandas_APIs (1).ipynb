```python
# ==========================================================
# PART 1: ETL on Messy Sales Data
# ==========================================================

# Activity 1 — Clean messy_sales_data.csv

# Upload messy_sales_data.csv to Colab Files panel,
# then run the cells below.
```

```python
# ==========================================================
# CELL 2 — EXTRACT: Load raw messy sales data
# ==========================================================

raw_df = pd.read_csv('messy_sales_data.csv')

# We store the RAW data in 'raw_df'

# ETL best practice: NEVER modify raw data

# Always work on a copy so you can go back to the original

print(f"Raw data loaded: {raw_df.shape[0]} rows, {raw_df.shape[1]} columns")

print(f"Columns: {raw_df.columns.tolist()}")

print("\nFirst 5 rows:")

raw_df.head()
```

```python
# ==========================================================
# CELL 3 — Diagnose ALL data quality problems before fixing
# ==========================================================

print("=" * 55)

print(" DATA QUALITY DIAGNOSIS REPORT ")

print("=" * 55)

# Check missing values

print("\nMissing Values:\n")

print(raw_df.isnull().sum())

# Check duplicate rows

print("\nDuplicate Rows:")

print(raw_df.duplicated().sum())

# Check data types

print("\nData Types:\n")

print(raw_df.dtypes)

# Unique category values

print("\nUnique Categories:\n")

print(raw_df['category'].unique())
```

```python
# ==========================================================
# CELL 4 — TRANSFORM: Create clean working copy
# ==========================================================

clean_df = raw_df.copy()

# We now work on clean_df
# raw_df remains untouched as backup

print("Working copy created successfully!")
```

```python
print(clean_df.columns)
```

```python
print(clean_df.columns)
```

```python
# ==========================================================
# CELL 5 — Handle Missing Values
# ==========================================================

print("Missing values before cleaning:\n")

print(clean_df.isnull().sum())

# Fill missing quantity values with median

clean_df['quantity'].fillna(
    clean_df['quantity'].median(),
    inplace=True
)

# Fill missing category values with mode

clean_df['category'].fillna(
    clean_df['category'].mode()[0],
    inplace=True
)

# Fill missing product values with mode

clean_df['product'].fillna(
    clean_df['product'].mode()[0],
    inplace=True
)

print("\nMissing values after cleaning:\n")

print(clean_df.isnull().sum())
```

```python
# ==========================================================
# CELL 6 — Remove Duplicate Rows
# ==========================================================

print("Rows before removing duplicates:")

print(clean_df.shape[0])

# Remove duplicate rows

clean_df.drop_duplicates(inplace=True)

print("\nRows after removing duplicates:")

print(clean_df.shape[0])
```

```python
# ==========================================================
# CELL 7 — Standardize Category Names
# ==========================================================

print("Unique categories before cleaning:\n")

print(clean_df['category'].unique())

# Convert category names to uppercase

clean_df['category'] = clean_df['category'].str.upper()

print("\nUnique categories after cleaning:\n")

print(clean_df['category'].unique())
```

```python
# ==========================================================
# CELL 8 — Final Clean Dataset Preview
# ==========================================================

print("Final cleaned dataset shape:")

print(clean_df.shape)

print("\nFirst 5 rows of cleaned data:\n")

clean_df.head()
```

```python
# ==========================================================
# CELL 9 — Save Cleaned Dataset
# ==========================================================

clean_df.to_csv('cleaned_sales_data.csv', index=False)

print("Cleaned dataset saved successfully!")
```

```python
# ==========================================================
# CELL 10 — Summary Report
# ==========================================================

print("=" * 55)

print(" FINAL ETL SUMMARY REPORT ")

print("=" * 55)

print(f"Total Rows : {clean_df.shape[0]}")

print(f"Total Columns : {clean_df.shape[1]}")

print("\nColumn Names:\n")

print(clean_df.columns.tolist())

print("\nData Types:\n")

print(clean_df.dtypes)

print("\nFirst 5 Rows of Cleaned Dataset:\n")

print(clean_df.head())
```

```python
# ==========================================================
# CHECK DUPLICATE ROWS
# ==========================================================

print(f"Before duplicate removal: {len(clean_df)} rows")

print(f"Duplicate rows: {clean_df.duplicated().sum()}")

# Show duplicate rows

print("\nDuplicate Rows:")

print(
    clean_df[
        clean_df.duplicated(
            subset=['order_id', 'customer_name'],
            keep=False
        )
    ]
)

# keep=False marks ALL duplicate rows

# ==========================================================
# CHECK DATE FORMAT
# ==========================================================

print("\nSample dates before parsing:")

print(clean_df['order_date'].head(8).tolist())

# ==========================================================
# CONVERT TO DATETIME
# ==========================================================

clean_df['order_date'] = pd.to_datetime(
    clean_df['order_date'],
    errors='coerce'
)

print("\nDates converted successfully!")

print(clean_df['order_date'].head())
```

```python
# ==========================================================
# STANDARDIZE CUSTOMER NAMES
# ==========================================================

print("Customer names before cleaning:\n")

print(clean_df['customer_name'].unique()[:10])

# Convert names to proper title case
# Remove extra spaces

clean_df['customer_name'] = (
    clean_df['customer_name']
    .str.strip()
    .str.title()
)

print("\nCustomer names after cleaning:\n")

print(clean_df['customer_name'].unique()[:10])
```
