````python
from groq import Groq

API_KEY = API_KEY = "YOUR_API_KEY"

client = Groq(api_key=API_KEY)

MODEL = "llama-3.3-70b-versatile"

print("Groq Connected Successfully")
````

````python
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {
            "role": "user",
            "content": "Hello"
        }
    ]
)

print(response.choices[0].message.content)
````

````python
def ask_llm(prompt):

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content
````

````python
answer = ask_llm("What is ETL?")

print(answer)
````

````python
prompt = """
Convert this employee record into JSON.

Input:
Ramesh Kumar,45000,Mumbai

Return ONLY valid JSON.

Example:
{
"name":"Ramesh Kumar",
"salary":45000,
"city":"Mumbai"
}
"""

result = ask_llm(prompt)

print(result)
````

````python
import json

clean_result = result.replace("```json", "").replace("```", "").strip()

parsed = json.loads(clean_result)

print(parsed)
````

````python
prompt = """
Convert this employee record into JSON.

Input:
Ramesh Kumar,45000,Mumbai

Return ONLY raw JSON.
Do NOT use markdown.
Do NOT use ```json.
"""
````

````python
import json

clean_result = result.strip()

parsed = json.loads(clean_result)

print(parsed)
print("Name:", parsed["name"])
print("Salary:", parsed["salary"])
print("Location:", parsed["location"])
````

````python
employees = [
    "Ramesh Kumar,45000,Mumbai",
    "Priya Nair,52000,Delhi",
    "Ananya Das,38000,Kolkata",
    "Arjun Singh,60000,Chennai",
    "Sneha Patel,55000,Pune"
]
````

````python
data = []

for emp in employees:
    parts = emp.split(",")

    data.append({
        "name": parts[0],
        "salary": int(parts[1]),
        "city": parts[2]
    })

print(data)
````

````python
import pandas as pd

df = pd.DataFrame(data)

print(df)
````

````python
print("Total Employees:", len(df))
print("Average Salary:", df["salary"].mean())
print("Highest Salary:", df["salary"].max())
print("Lowest Salary:", df["salary"].min())
````

````python
import matplotlib.pyplot as plt

plt.bar(df["name"], df["salary"])
plt.xticks(rotation=45)
plt.title("Employee Salary Analysis")
plt.show()
````
