```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/student_performance.csv')

df.head()
```

```python
df['total_score'] = df['math_score'] + df['programming_score']
```

```python
conn = sqlite3.connect('student.db')

df.to_sql('students', conn, if_exists='replace', index=False)

print("Database connected successfully!")
```

```python
# Panel 1 data: avg math by department
query1 = """
SELECT department,
ROUND(AVG(math_score),2) AS avg_math
FROM students
GROUP BY department;
"""

avg_math = pd.read_sql_query(query1, conn)

# Panel 2 data: student count by department
query2 = """
SELECT department,
COUNT(*) AS student_count
FROM students
GROUP BY department;
"""

student_count = pd.read_sql_query(query2, conn)

# Panel 3 data: top 8 students by total score
query3 = """
SELECT name,
total_score
FROM students
ORDER BY total_score DESC
LIMIT 8;
"""

top_students = pd.read_sql_query(query3, conn)

# Panel 4 data: gender-wise average attendance
query4 = """
SELECT gender,
ROUND(AVG(attendance_percentage),2) AS avg_attendance
FROM students
GROUP BY gender;
"""

gender_attendance = pd.read_sql_query(query4, conn)

print("SQL Queries Executed Successfully!")
```

```python
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# ── Panel 1 (top-left): Avg Math by Department - BAR Chart ──
axs[0,0].bar(avg_math['department'], avg_math['avg_math'])

axs[0,0].set_title('Average Math Score by Department')
axs[0,0].set_xlabel('Department')
axs[0,0].set_ylabel('Average Math Score')

# ── Panel 2 (top-right): Student Count Pie Chart ──
axs[0,1].pie(
    student_count['student_count'],
    labels=student_count['department'],
    autopct='%1.1f%%'
)

axs[0,1].set_title('Student Count by Department')

# ── Panel 3 (bottom-left): Top 8 Students Horizontal Bar ──
axs[1,0].barh(
    top_students['name'],
    top_students['total_score']
)

axs[1,0].set_title('Top 8 Students by Total Score')
axs[1,0].set_xlabel('Total Score')

# ── Panel 4 (bottom-right): Gender Avg Attendance Bar ──
axs[1,1].bar(
    gender_attendance['gender'],
    gender_attendance['avg_attendance']
)

axs[1,1].set_title('Average Attendance by Gender')
axs[1,1].set_xlabel('Gender')
axs[1,1].set_ylabel('Attendance')

plt.tight_layout()

plt.show()
```

## Question 2
What is the difference between WHERE and HAVING?

WHERE:
Used to filter rows before grouping.

Example:

Used to filter groups after GROUP BY.

Example:
SELECT department, AVG(math_score)

```python
# ==========================================================
# STUDENT PERFORMANCE DASHBOARD
# ==========================================================

# Import Libraries
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv('/content/student_performance.csv')

# Create Total Score Column
df['total_score'] = (
    df['math_score'] +
    df['science_score'] +
    df['english_score'] +
    df['programming_score']
)

# Connect SQLite Database
conn = sqlite3.connect('student.db')

# Store Dataset into SQL Table
df.to_sql('students', conn, if_exists='replace', index=False)

# ==========================================================
# SQL QUERY 1
# Average Math Score by Department
# ==========================================================

query1 = """
SELECT department,
ROUND(AVG(math_score),1) AS avg_math
FROM students
GROUP BY department
ORDER BY avg_math DESC;
"""

avg_math = pd.read_sql_query(query1, conn)

# ==========================================================
# SQL QUERY 2
# Student Count by Department
# ==========================================================

query2 = """
SELECT department,
COUNT(*) AS student_count
FROM students
GROUP BY department;
"""

student_count = pd.read_sql_query(query2, conn)

# ==========================================================
# SQL QUERY 3
# Top 8 Students by Total Score
# ==========================================================

query3 = """
SELECT name,
total_score
FROM students
ORDER BY total_score DESC
LIMIT 8;
"""

top_students = pd.read_sql_query(query3, conn)

# ==========================================================
# SQL QUERY 4
# Average Attendance by Gender
# ==========================================================

query4 = """
SELECT gender,
ROUND(AVG(attendance_percentage),1) AS avg_attendance
FROM students
GROUP BY gender;
"""

gender_attendance = pd.read_sql_query(query4, conn)

# ==========================================================
# CREATE DASHBOARD
# ==========================================================

fig, axs = plt.subplots(2, 2, figsize=(16, 12))

# ==========================================================
# PANEL 1
# Avg Math by Department
# ==========================================================

bars1 = axs[0,0].bar(
    avg_math['department'],
    avg_math['avg_math'],
    color=['#4CAF50','#2196F3','#FF9800']
)

axs[0,0].set_title(
    'Average Math Score by Department',
    fontsize=14,
    fontweight='bold'
)

axs[0,0].set_ylabel('Average Score')
axs[0,0].set_ylim(0, 105)

axs[0,0].grid(axis='y', linestyle='--', alpha=0.5)

# Value labels
for bar in bars1:
    height = bar.get_height()

    axs[0,0].text(
        bar.get_x() + bar.get_width()/2,
        height + 1,
        f'{height}',
        ha='center',
        fontsize=10
    )

# ==========================================================
# PANEL 2
# Student Count Pie Chart
# ==========================================================

axs[0,1].pie(
    student_count['student_count'],
    labels=student_count['department'],
    autopct='%1.1f%%',
    colors=['#42A5F5','#66BB6A','#FFA726'],
    wedgeprops={'edgecolor':'white'}
)

axs[0,1].set_title(
    'Student Count by Department',
    fontsize=14,
    fontweight='bold'
)

# ==========================================================
# PANEL 3
# Top 8 Students Horizontal Bar
# ==========================================================

bars3 = axs[1,0].barh(
    top_students['name'],
    top_students['total_score'],
    color='#AB47BC'
)

axs[1,0].invert_yaxis()

axs[1,0].set_title(
    'Top 8 Students by Total Score',
    fontsize=14,
    fontweight='bold'
)

axs[1,0].set_xlabel('Total Score')

axs[1,0].grid(axis='x', linestyle='--', alpha=0.5)

# Value labels inside bars
for bar in bars3:
    width = bar.get_width()

    axs[1,0].text(
        width - 10,
        bar.get_y() + bar.get_height()/2,
        f'{width}',
        va='center',
        color='white',
        fontsize=9
    )

# ==========================================================
# PANEL 4
# Gender Avg Attendance
# ==========================================================

bars4 = axs[1,1].bar(
    gender_attendance['gender'],
    gender_attendance['avg_attendance'],
    color=['#EC407A','#29B6F6']
)

axs[1,1].set_title(
    'Average Attendance by Gender',
    fontsize=14,
    fontweight='bold'
)

axs[1,1].set_ylabel('Attendance Percentage')

axs[1,1].set_ylim(0, 105)

axs[1,1].grid(axis='y', linestyle='--', alpha=0.5)

# Value labels
for bar in bars4:
    height = bar.get_height()

    axs[1,1].text(
        bar.get_x() + bar.get_width()/2,
        height + 1,
        f'{height}%',
        ha='center',
        fontsize=10
    )

# ==========================================================
# DASHBOARD TITLE
# ==========================================================

fig.suptitle(
    'STUDENT PERFORMANCE DASHBOARD',
    fontsize=20,
    fontweight='bold'
)

# Adjust Layout
plt.tight_layout(rect=[0, 0, 1, 0.96])

# ==========================================================
# SAVE DASHBOARD
# ==========================================================

plt.savefig(
    'student_dashboard.png',
    dpi=150,
    bbox_inches='tight'
)

# Show Dashboard
plt.show()

# Confirmation Message
print("Dashboard saved successfully as student_dashboard.png")

# Close Database
conn.close()
```
