```python
import pandas as pd

# Load Titanic Dataset
df = pd.read_csv('/content/Titanic-Dataset.csv')

# Display First 5 Rows
print("="*50)
print("TITANIC DATASET ANALYSIS")
print("="*50)

print("\nFIRST 5 ROWS")
print(df.head())

# Dataset Information
print("\nDATASET INFO")
print("-"*30)
print(df.info())

# Dataset Shape
print("\nDATASET SHAPE")
print("-"*30)
print(df.shape)

# Column Names
print("\nCOLUMN NAMES")
print("-"*30)
print(df.columns)

# Statistical Summary
print("\nSTATISTICAL SUMMARY")
print("-"*30)
print(df.describe())

# Missing Values
print("\nMISSING VALUES")
print("-"*30)
print(df.isnull().sum())

# Survival Count
print("\nSURVIVAL COUNT")
print("-"*30)
print(df['Survived'].value_counts())

# Gender Count
print("\nGENDER COUNT")
print("-"*30)
print(df['Sex'].value_counts())

# Average Age
print("\nAVERAGE AGE")
print("-"*30)
print(df['Age'].mean())

# Top 5 Oldest Passengers
print("\nTOP 5 OLDEST PASSENGERS")
print("-"*30)

oldest = df.sort_values(by='Age', ascending=False).head(5)

print(oldest[['Name','Age']])

# Passenger Class Count
print("\nPASSENGER CLASS COUNT")
print("-"*30)
print(df['Pclass'].value_counts())
```
