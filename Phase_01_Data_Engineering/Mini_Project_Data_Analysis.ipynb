```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv('/content/student_performance.csv')

# Project Title
print("="*50)
print("      STUDENT PERFORMANCE EXPLORER")
print("="*50)

# Dataset Overview
print("\nDATASET OVERVIEW")
print("-"*30)

print("Total number of students:", df.shape[0])
print("Total number of columns:", df.shape[1])
print("Number of unique departments:", df['department'].nunique())

# Department-wise Analysis
print("\nDEPARTMENT-WISE ANALYSIS")
print("-"*30)

print(df['department'].value_counts())

# Score Analysis
print("\nSCORE ANALYSIS")
print("-"*30)

print("Highest Math Score:", df['math_score'].max())
print("Lowest Math Score:", df['math_score'].min())
print("Average Math Score:", round(df['math_score'].mean(),2))
print("Average Programming Score:", round(df['programming_score'].mean(),2))
print("Average Attendance Percentage:", round(df['attendance_percentage'].mean(),2))

# Create Total Score
df['total_score'] = df['math_score'] + df['programming_score']

# Gender-wise Analysis
print("\nGENDER-WISE ANALYSIS")
print("-"*30)

gender_avg = df.groupby('gender')['total_score'].mean().round(2)

print(gender_avg)

# Top 5 Students
print("\nTOP 5 STUDENTS")
print("-"*30)

top_students = df.sort_values(by='total_score', ascending=False).head(5)

print(top_students[['name','department','total_score']])

# Attendance Analysis
print("\nATTENDANCE ANALYSIS")
print("-"*30)

low_attendance = df[df['attendance_percentage'] < 75]

if low_attendance.empty:
    print("All students have attendance above 75%")
else:
    print(low_attendance[['name','department','attendance_percentage']])

# Graph
print("\nGRAPH VISUALIZATION")

df['department'].value_counts().plot(kind='bar')

plt.title("Department-wise Student Count")
plt.xlabel("Department")
plt.ylabel("Count")

plt.show()
```
